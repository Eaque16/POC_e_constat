```python
import subprocess, torch
print(subprocess.check_output(['nvidia-smi', '--query-gpu=name,memory.total', '--format=csv,noheader'], text=True))
assert torch.cuda.is_available(), 'Activez le GPU dans les paramètres Kaggle.'
```

```python
from pathlib import Path
AUDIO_PATH = Path('/kaggle/input/VOTRE_DATASET/audio.wav')
assert AUDIO_PATH.exists(), 'Ajoutez un Dataset Kaggle contenant un WAV de démonstration.'
```

```python
import time
from faster_whisper import WhisperModel
MODEL = 'deepdml/faster-whisper-large-v3-turbo-ct2'
started = time.perf_counter()
model = WhisperModel(MODEL, device='cuda', compute_type='float16')
load_seconds = time.perf_counter() - started
started = time.perf_counter()
segments, info = model.transcribe(str(AUDIO_PATH), language='fr', beam_size=1, vad_filter=True)
segments = list(segments)
elapsed = time.perf_counter() - started
text = ' '.join(segment.text.strip() for segment in segments)
print({'model': MODEL, 'load_s': round(load_seconds, 3), 'inference_s': round(elapsed, 3), 'audio_s': round(info.duration, 3), 'rtf': round(elapsed / max(info.duration, .01), 3)})
print(text)
```
